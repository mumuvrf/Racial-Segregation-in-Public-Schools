```python
import pandas as pd
```

```python
df = pd.read_csv('../data/public_school_data.csv', skiprows=6)
df.head(3)
```

```python
schools_info = df.loc[:, 
    ["School ID (12-digit) - NCES Assigned [Public School] Latest available year",
    "School Name", "State Name [Public School] Latest available year"]
]

schools_columns = {
    "School ID (12-digit) - NCES Assigned [Public School] Latest available year" : "school_id",
    "School Name": "school_name",
    "State Name [Public School] Latest available year": "state_name"
}

schools_info = schools_info.rename(columns=schools_columns)
schools_info['school_id'] = schools_info['school_id'].astype(str)
schools_info.head(2)
```

```python
school_years = [
    '2008-09', '2009-10', '2010-11', '2011-12',
    '2012-13', '2013-14', '2014-15', '2015-16',
    '2016-17', '2017-18', '2018-19', '2019-20',
    '2020-21', '2021-22', '2022-23', '2023-24',
    '2024-25'
]

def data_transformer(row, year: str = None):
    new_row = {
        "school_id": row["School ID (12-digit) - NCES Assigned [Public School] Latest available year"],
        "year": year,
        "white_students": row[f"White Students [Public School] {year}"],
        "black_students": row[f"Black or African American Students [Public School] {year}"],
        "total_students": row[f"Total Race/Ethnicity [Public School] {year}"],
        "free_and_reduced_lunch": row[f"Free and Reduced Lunch Students [Public School] {year}"]
    }
    return new_row

racial_data = pd.DataFrame(
    data_transformer(row, year)
    for _, row in df.iterrows()
    for year in school_years
)
racial_data["school_id"] = racial_data["school_id"].astype(str)
racial_data.head(2)
```
